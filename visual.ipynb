```python
import matplotlib.pyplot as mp
import seaborn as sb
import pandas as pd

data=pd.read_csv("cleaned_drone_data.csv")
data.columns
```

```python
ax=data["Drone Model"].value_counts().plot(kind="bar",figsize=(16,6),color="cyan",edgecolor="black",linewidth=1.5)

ax.set_title("Drone Model VS Flight Count")
ax.set_ylabel("Flight Count",fontsize=12)
ax.set_xlabel("Drone Model",fontsize=12)
mp.pyplot.tight_layout()
mp.pyplot.show()

# highest flight count-SnapShot Mini (114)
# lowest flight count-Inspecta X (2)
```

```python
ax=data.plot(x="Flight Duration (minutes)",y="Distance Flown (km)",kind="scatter",s=35,color="skyblue",alpha=0.6,figsize=(16,6))
ax.set_title("Duration VS Distance")
ax.set_xlabel("Flight Duration (minutes)")
ax.set_ylabel("Distance Flown (km)")
mp.pyplot.tight_layout()
mp.pyplot.show()

# mostly longer flights cover greater distances. mostly positive relationship
```

```python
ax=data.boxplot(column="Wind Speed (m/s)",by="Altitude (meters)",figsize=(16,6),flierprops={'markersize':3})

ax.set_xlabel("Altitude")
ax.set_ylabel("Wind Speed (m/s)")
ax.set_title("Wind Speed VS Altitude")
ax.set_xticks(ax.get_xticks()[::5])
mp.pyplot.xticks(rotation=45,fontsize=8)
mp.pyplot.tight_layout()
mp.pyplot.show()

# at low altitudes, wind speeds are generally low and stable until 25m after which it jumps suddenly
# between 35m to 95m, wind speeds are volatile (many outliers)
```

```python
ax=data["Actual Carry Weight (kg)"].value_counts().plot(kind="hist",bins=15,figsize=(16,6),color="blue")
ax.set_title("Distribution of Actual Carry Weight")
ax.set_xlabel("Actual Carry Weight (kg)")
ax.set_ylabel("FLight Count")
mp.pyplot.show()

# majority of  flights cary less load (1-3 kg)
# few flights in outlier group carrying about 24kg
```

```python
ax=data["Flight Date"].value_counts().plot(kind="line",figsize=(16,6),color="red")
ax.set_title("Flight Date Trend",fontweight="bold")
ax.set_xlabel("Flight Date",fontweight="bold")
ax.set_ylabel("Number of Flights",fontweight="bold")
mp.pyplot.tight_layout()
mp.pyplot.show()

# flight frequency peaked around april-may 2025
# flight frequency dipped to a steady line till 2026
```
